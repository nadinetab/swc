#### RBSPA_REL04_ECT-MAGEIS-L3@0

##### Datasets
We have `RBSPA_REL04_ECT-MAGEIS-L3@0` and `RBSPA_REL04_ECT-MAGEIS-L3@1` available to us. I believe @0 this deals with electron data, @1 corresponds to protons.

```python
server = "https://cdaweb.gsfc.nasa.gov/hapi/"
dataset = "RBSPA_REL04_ECT-MAGEIS-L3@0"
start = "2013-06-13T00:00:00Z"
stop = "2013-06-14T00:00:00Z"
parameters = (
    ""  # Passing in an empty string should give all of the available parameters :)
)

data, meta = hapi(server, dataset, parameters, start, stop)
```

```python
print(type(meta))  # dictionary
meta
```

```python
print(type(data))  # numpy array
print(data.shape)
data
```

```python
data[0]
```

```python
# How many parameters?
print("Num of parameters: ", len(meta["parameters"]))

# Names and descriptions
for p in meta["parameters"]:
    try:
        print(p["name"], ",", p["size"], ", description: ", p["description"])
    except KeyError:
        print(p["name"])
```

FEDU_plasmagram and its error measurements are probably the ones we're most interested in, alongside the magnetic coordinates (L_star/L/B_eq/MLT, etc)

```python
print(type(data))
print(data.dtype)
print(data.shape)
print(data.dtype.names)
```
