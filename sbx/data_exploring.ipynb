# Hapi access and dataset exploration

```python
from hapiclient import hapi
help(hapi)
```

```python
server = "https://cdaweb.gsfc.nasa.gov/hapi/"
dataset = "RBSPA_REL04_ECT-MAGEIS-L3@0"
start = "2012-06-13T00:00:00"
stop = "2012-06-14T00:00:00"
parameters = "" # Passing in an empty string should give all of the available parameters :)

data, meta = hapi(server, dataset, parameters, start, stop)
```

```python
print(type(meta))
meta # I think this is a dictionary. So says hapi_demo.ipynb as well.
```

```python
# How many parameters?
print('Num of parameters: ', len(meta['parameters']))
```
